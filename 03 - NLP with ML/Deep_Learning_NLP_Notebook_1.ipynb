## Removing Punctuation and Numbers

w.isalpha(), is a method that checks if the token contains only alphabetic characters (letters) and returns True if it does.
If w consists only of letters (i.e., does not contain any punctuation marks or numbers), this token is included in the tokens_without_punc list.


w.isalnum(), is a method that checks if the token contains only alphanumeric characters (letters or numbers) and returns True if it does.
If w consists only of letters or numbers, this token is included in the tokens_without_punc list.
Therefore, after running this code, the tokens_without_punc list will only contain tokens made up of letters or numbers from the word_token list. Tokens containing punctuation marks or other special characters will not be included in this list.

```python
tokens_without_punc = [w for w in word_token if w.isalpha()]   # list comprehension
tokens_without_punc  # punctuation, special characters, and numbers are removed. If numbers are to be kept, use isalnum().
```

With this operation, a cleaned word list is obtained where characters like punctuation marks and numbers are removed. This can be useful in text analysis or natural language processing tasks when only words are of interest, or when numbers and special characters are not considered.


```python
tokens_without_punc = [w for w in word_token if w.isalnum()]        # isalnum() for numbers and objects
tokens_without_punc          # punctuation and special characters are removed. If numbers are to be kept, use isalnum().
```

When using the **WordNetLemmatizer().lemmatize()** function, the **pos** (part of speech) argument specifies which word class (noun, verb, adjective, etc.) the word belongs to. This provides the grammatical information needed for lemmatization. Here are some values that can be used for the **pos** argument:

- 'n': Noun (for nouns)
- 'v': Verb (for verbs)
- 'a': Adjective (for adjectives)
- 'r': Adverb (for adverbs)

For example, to lemmatize the word "runs" as a verb, you can use **pos='v'**:


The expression **WordNetLemmatizer().lemmatize("runs", pos='v')** performs the lemmatization of the word "runs" according to the "v" (verb) part of speech.

Lemmatization is the process of finding the root (lemma) of a word. It aims to obtain the basic meaning or root of the word by considering its different inflections or forms. In this case, since the word "runs" is a verb, it is specified with the parameter **pos='v'**.

As output, "run" is obtained. Since "runs" is a verb, lemmatization converts it to the word "run."

## Cleaning Function - for classification (NOT for sentiment analysis)

```python
def cleaning(data):
    
    #1. Tokenize
    text_tokens = word_tokenize(data.lower()) 
    
    #2. Remove Puncs
    tokens_without_punc = [w for w in text_tokens if w.isalpha()]
    
    #3. Removing Stopwords
    tokens_without_sw = [t for t in tokens_without_punc if t not in stop_words]
    
    #4. lemma
    text_cleaned = [WordNetLemmatizer().lemmatize(t) for t in tokens_without_sw]
    
    #joining
    return " ".join(text_cleaned)
```

## CountVectorization

```python
X = df2["text"]
y = df2["airline_sentiment"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, stratify = y, random_state = 42)
```

The **stratify = y** parameter has been used. This ensures that the class ratios in the **y** variable are preserved in both the training set and the test set. This allows the model to be trained and tested in a more balanced manner.

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
vectorizer = CountVectorizer()                      # Similar to what we do in scaling in ML.
X_train_count = vectorizer.fit_transform(X_train)   # On fit, it detects all unique words, and on transform, it counts them.
X_test_count = vectorizer.transform(X_test)         # If a word in X_test is not present in X_train, it will not be considered.
                                                    # Therefore, it needs to be trained on a large corpus to get good results...
```

Here, you are using **CountVectorizer** to convert the data into a count vector (bag of words). **CountVectorizer** is a commonly used method to transform text data into word-level count vectors.

The data transformation process occurs in two steps:

- **fit_transform(X_train)**: In this step, you train the **CountVectorizer** on the training data (**X_train**). This process identifies all unique words and creates a count matrix where each word is represented by a column. Each cell contains the frequency of a word's occurrence.

- **transform(X_test)**: In this step, you apply the same transformation to the test data (**X_test**). However, it only counts the words that were seen in the training data and transforms these word counts into a count matrix for the test data. In other words, a count matrix is created to represent the test data based on the vocabulary learned from the training data.

As a result, the **X_train_count** and **X_test_count** matrices represent the process of converting text data into count vectors using the bag of words representation. This representation is used to process text data numerically and feed it into machine learning models.

```python
vectorizer.get_feature_names_out()            # gives all the unique tokens
```

```python
X_train_count.toarray()   
```

When the **toarray()** method is called on a variable named **X_train_count**, this method converts the sparse matrix in the **X_train_count** variable into a dense matrix and returns the result as a NumPy array. This conversion creates a dense matrix where the values are explicitly represented along with their row and column indices, rather than using the compact representation of a sparse matrix.
