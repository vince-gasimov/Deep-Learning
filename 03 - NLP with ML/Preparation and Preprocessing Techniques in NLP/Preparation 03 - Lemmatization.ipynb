### Wordnet Lemmatizer

Lemmatization is a normalization technique in nlp. The output we will get after lemmatization is called ‘lemma’. After lemmatization, we will be getting a valid word that means the same thing.

NLTK provides WordNetLemmatizer class which is a thin wrapper around the wordnet corpus. This class uses morphy() function to the WordNet CorpusReader class to find a lemma. Let us understand it with an example −

```python
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
lemmatizer.lemmatize("going")
```

```python
lemmatizer.lemmatize("going", pos = 'v')
```

```python
lemmatizer.lemmatize("goes", pos = 'v')
```

```python
'''
POS = Parts of Speech
n --> Noun
v --> Verb
a --> adjective
r --> adverb
'''
```

```python
words=["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalized"]
```

```python
for word in words:
    print(word+"---->"+lemmatizer.lemmatize(word,pos='v'))
    # by default it sets the pos as n, and first try without pos
```

```python
lemmatizer.lemmatize("fair",pos='r'),lemmatizer.lemmatize("sportingly")
```
