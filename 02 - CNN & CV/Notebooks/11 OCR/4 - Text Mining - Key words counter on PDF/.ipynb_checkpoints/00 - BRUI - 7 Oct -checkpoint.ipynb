```python
# Lower Case donusturdu, 2'li 3'luleri de yakaliyor; Brexit kavram sayisi artirildi; Sadece gerekli index kelimelerini veriyor
# Sonuclar idare eder - 6 Oct 18.00 "0" tane olan kelimeler atildi. KELIMELER AZALTILDI!!!

# Buradan cikan sonucu "indexes" kelime listesi ile karsilastirdim, mukemmel tuttu. PERFECT dedigimin kodu bu. 

import spacy
import re
import os
import fitz  # PyMuPDF
import pandas as pd

# Load spaCy language model
nlp = spacy.load("en_core_web_lg")

# Define uncertainty terms (exact words, no lemmatization)
uncertainty_related_words = ["fear", "indecision", "instability", "jittery", "nervousness", "precarious", "tense", 
                             "tension", "uncertain", "uncertainly", "uncertainty", "unclear", "unknown", "unpredictable", 
                             "unsettled", "unstable", "volatile", "volatility", "worry"]

# Define Brexit-related phrases and words - exact phrases only
brexit_related_phrases = ["article 50", "brexit", "brexit-related", "customs union", "eu exit", "eu membership", 
                          "eu withdrawal", "exit deal", "exit from the eu", "exit the eu", "exit time", 
                          "exiting", "exiting the eu", "exiting the european union", "free movement",  
                          "internal market bill", "leave the eu",  "northern ireland protocol", 
                          "post-brexit", "pre-brexit", "referendum", "regulatory alignment", "regulatory framework", 
                          "single market", "trade arrangement", "trade negotiations", "transition period", "uk exits", 
                          "uk-eu relations", "uk-eu trade deal", "uk's withdrawal", "withdrawal agreement", 
                          "withdrawal from the eu"]

# Define Covid-19 related phrases and words - exact phrases only
covid_related_phrases = ["coronavirus", "covid", "covid-19", "lockdown", "outbreak", "pandemic", "quarantine", 
                         "vaccination", "vaccine"]

# Define words to ignore uncertainty count
ignore_words = ["scotland", "scottish"]

# Context window size in terms of number of words
context_window_size = 10  # Increased to improve context capture

# Folder path containing PDF reports
folder_path = r"C:\Users\ismet\OneDrive - University of Strathclyde\Desktop\Unv Project 1 Oct\BRUI\Reports"

# Create a list to store results
results_list = []

# Helper function to generate n-grams (bi-grams, tri-grams)
def generate_ngrams(words, n):
    return [' '.join(words[i:i + n]) for i in range(len(words) - n + 1)]

# Iterate through files in the folder
for filename in os.listdir(folder_path):
    if filename.endswith(".pdf"):
        file_path = os.path.join(folder_path, filename)

        with fitz.open(file_path) as pdf_file:
            text = ""
            for page in pdf_file:
                text += page.get_text()

            # Preprocess text for cleaner analysis
            text = re.sub(r'\s+', ' ', text)  # Combine multiple spaces
            text = re.sub(r'\n', ' ', text)   # Replace newlines with spaces
            text = ''.join(ch for ch in text if ch.isprintable())  # Remove non-printable characters
            text = text.lower()  # Convert entire text to lowercase

            # Get total word count of the document
            total_word_count = len(text.split())

            # Analyze paragraphs for uncertainty mentions
            paragraphs = text.split("\n\n")
            brexit_uncertainty_count = 0
            covid_uncertainty_count = 0
            both_uncertainty_count = 0
            total_uncertainty_count = 0

            for paragraph in paragraphs:

                # Use spaCy for efficient named entity recognition (NER)
                doc = nlp(paragraph)

                # Remove stop words from the document and make lowercase
                doc_words = [token.text for token in doc if not token.is_stop and token.is_alpha]
                
                # Generate bi-grams and tri-grams
                bigrams = generate_ngrams(doc_words, 2)
                trigrams = generate_ngrams(doc_words, 3)

                # Combine all tokens, bi-grams, and tri-grams into a single list
                all_tokens = doc_words + bigrams + trigrams

                # Count uncertainty occurrences within context window
                for i, token in enumerate(all_tokens):
                    # Check if the token is in the uncertainty-related words (no lemmatization)
                    if token in uncertainty_related_words:
                        # Define the context window by word indices
                        start_index = max(0, i - context_window_size)
                        end_index = min(len(all_tokens), i + context_window_size + 1)
                        context_window = ' '.join(all_tokens[start_index:end_index])

                        # Use regex-based search for partial matches to improve capture
                        brexit_in_context = any(re.search(r'\b' + re.escape(phrase.lower()) + r'\b', context_window.lower()) for phrase in brexit_related_phrases)
                        covid_in_context = any(re.search(r'\b' + re.escape(phrase.lower()) + r'\b', context_window.lower()) for phrase in covid_related_phrases)
                        ignore_in_context = any(ignore.lower() in context_window.lower() for ignore in ignore_words)

                        # If "referendum" is in context and "Scotland" or "Scottish" is in context, skip counting
                        if "referendum" in context_window.lower() and ignore_in_context:
                            continue

                        total_uncertainty_count += 1
                        if brexit_in_context and not covid_in_context:
                            brexit_uncertainty_count += 1
                        elif covid_in_context and not brexit_in_context:
                            covid_uncertainty_count += 1
                        elif brexit_in_context and covid_in_context:
                            both_uncertainty_count += 1

            # Add results to the list
            results_list.append({
                "File": filename,
                "Total Words Number": total_word_count,
                "Total Uncertainty": total_uncertainty_count,
                "Brexit Related Uncertainty": brexit_uncertainty_count,
                "Covid Related Uncertainty": covid_uncertainty_count,
                "Brexit & Covid Related Uncertainty": both_uncertainty_count
            })

# Convert results list to DataFrame
results = pd.DataFrame(results_list)

# Get current working directory
current_dir = os.getcwd()

# Construct output file path
excel_file_path = os.path.join(current_dir, "7_Oct_Indexes_11.56.xlsx")

# Save results to Excel file
results.to_excel(excel_file_path, index=False)

print("Excel file successfully created:", excel_file_path)
```
